Após analise do dataset acima quais incompatibilidades ou problemas de qualidade de dados voce identificou? Quais colunas tem valores faltantes? Quais colunas tem valores inconsistentes? Quais colunas tem valores duplicados? Quais colunas tem valores fora do intervalo esperado?

Uma maneira inteligente de fazer isso é talvez fazer uma coluna sumária que analise cada coluna e sua completude, consistência, duplicidade e validade. Portanto crie uma tabela sumária sobre com as colunas descritas abaixo:

```python
#nome da coluna
#tipo de dados
#quantidade de valores únicos
#quantidade de valores faltantes
#quantidade de valores duplicados
#quantidade de dados inteiros
#quantidade de dados float
#quantidade de dados string
#quantidade de dados booleanos
#quantidade de dados nulos

#criando a tabela
summary = pd.DataFrame({
    'Column': ds.columns,
    'DataType': ds.dtypes,
    'UniqueValues': ds.nunique(),
    'MissingValues': ds.isnull().sum(),
    'DuplicatedValues': ds.apply(lambda x: x.duplicated().sum()),
    'IntegerValues': ds.apply(lambda x: x.apply(lambda y: isinstance(y, int)).sum()),
    'FloatValues': ds.apply(lambda x: x.apply(lambda y: isinstance(y, float)).sum()),
    'StringValues': ds.apply(lambda x: x.apply(lambda y: isinstance(y, str)).sum()),
    'BooleanValues': ds.apply(lambda x: x.apply(lambda y: isinstance(y, bool)).sum()),
})

summary
```

```python
#duplicated values in columns "PassengerId" and "Name"
print("Valores duplicados na coluna 'PassengerId':")
print(ds['PassengerId'].duplicated().sum())
print("Valores duplicados na coluna 'Name':")
print(ds['Name'].duplicated().sum())
```

```python
#qual o padrao dos valores de ticket?
print("Valores únicos na coluna 'Ticket':")
print(ds['Ticket'].unique())
```

```python
#valores unicos dos tickets primeira classe
print("Valores únicos na coluna 'Ticket' para passageiros da primeira classe:")
print(ds[ds['Pclass'] == 1]['Ticket'].unique())
```

```python
#ha valores duplicados de ticket ?
print("Valores duplicados na coluna 'Ticket':")
print(ds['Ticket'].duplicated().sum())
```

```python
#print as rows with duplicated tickets
print("Linhas com valores duplicados na coluna 'Ticket':")
print(ds[ds['Ticket'].duplicated(keep=False)])
```

```python
#Qual era a proporção de sobreviventes em relação ao total de passageiros?
total_passengers = len(ds)
survivors = ds['Survived'].sum()
proportion_survivors = survivors / total_passengers
print(f"Proporção de sobreviventes: {proportion_survivors:.2%}")
```

```python
#Qual era a proporção de sobreviventes por classe de cabine?
survivors_by_class = ds.groupby('Pclass')['Survived'].mean()
print("Proporção de sobreviventes por classe de cabine:")
print(survivors_by_class)
```

```python
#Qual era a proporção de sobreviventes por sexo?
survivors_by_sex = ds.groupby('Sex')['Survived'].mean()
print("Proporção de sobreviventes por sexo:")
print(survivors_by_sex)
```

```python
#Qual era a proporção de sobreviventes por faixa etária?
# Criar faixas etárias
bins = [0, 12, 18, 35, 60, 100]
labels = ['0-12', '13-18', '19-35', '36-60', '60+']
ds['AgeGroup'] = pd.cut(ds['Age'], bins=bins, labels=labels)
survivors_by_age_group = ds.groupby('AgeGroup')['Survived'].mean()
print("Proporção de sobreviventes por faixa etária:")
print(survivors_by_age_group)
```

```python
#A coluna "SibSp" indica o número de irmãos/cônjuges a bordo.
#A coluna "Parch" indica o número de pais/filhos a bordo.

#qual das duas colunas usar?
#Se quero analisar a proporção de sobreviventes por número de irmãos/cônjuges a bordo, devo usar a coluna "SibSp". Se quero analisar a proporção de sobreviventes por número de pais/filhos a bordo, devo usar a coluna "Parch".
```

```python
#unique values in SibSp column
print("Valores únicos na coluna 'SibSp':")
print(ds['SibSp'].unique())

#portanto numero inteiro de irmãos/cônjuges a bordo, variando de 0 a 8.
```

```python
# Qual era a proporção de sobreviventes por número de irmãos/cônjuges a bordo?
survivors_by_siblings = ds.groupby('SibSp')['Survived'].mean()
print("Proporção de sobreviventes por número de irmãos/cônjuges a bordo:")
print(survivors_by_siblings)
```

```python
#Qual era a proporção de sobreviventes por número de pais/filhos a bordo?
survivors_by_parents = ds.groupby('Parch')['Survived'].mean()
print("Proporção de sobreviventes por número de pais/filhos a bordo:")
print(survivors_by_parents)
```

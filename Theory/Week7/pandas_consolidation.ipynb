## Series vs DataFrame

No **pandas**, existem duas estruturas de dados fundamentais: **Series** e **DataFrame**. Compreender bem a diferença entre estas duas estruturas é essencial, pois praticamente todas as operações que realizamos vão envolver uma delas.

### Series

Uma **Series** é uma estrutura de dados **unidimensional**, ou seja, funciona como uma lista ou uma coluna de uma tabela.

- Possui **um único eixo** (valores)
- Cada elemento está associado a um **índice**
- Pode armazenar qualquer tipo de dados (inteiros, strings, floats, etc.)

```python
obj2 = pd.Series([4, 7,-5, 3], index=["d", "b", "a", "c"])
obj2
```

```python
#selecionar uma linha passando o nome do indice
print(obj2["a"])
```

```python
#iloc funciona
print(obj2.iloc[2])
```

```python
#loc também
print(obj2.loc["c"])
```

```python
#filtros em serie de dados. Aqui nao precisa indicar a coluna. Só tem uma coluna
obj2[obj2 > 0]
```

```python
#operacoes com todos os elementos da serie
print(obj2 * 2)
```

```python
#quando iterage com a serie o que acontece
for i in obj2:
    print(i)

#quando iteragem com um dataframe o que acontecia??
```

```python
#operador in aqui testa um index e nao um valor
"b" in obj2
```

```python
#criando uma serie a partir de um dicionário
sdata = {"Ohio": 35000, "Texas": 71000, "Oregon": 16000, "Utah": 5000}
obj3 = pd.Series(sdata)
obj3
#portanto nao precisa criar como fizemos antes, a partir de duas listas. Awui ele ja identifica as chaves do dicionário como indices e os valores como os dados da serie
```

### DataFrame

Um **DataFrame** é uma estrutura **bidimensional**, ou seja, uma tabela com linhas e colunas.

- Possui **dois eixos** (linhas e colunas)
- Cada coluna é, na prática, uma **Series**
- Permite armazenar diferentes tipos de dados em colunas diferentes

## O Index no Pandas

O **Index** é um dos elementos fundamentais no pandas e está presente tanto em **Series** quanto em **DataFrames**. Ele funciona como um conjunto de **rótulos (labels)** que identificam cada linha (e também cada coluna, no caso de DataFrames).

### O que é o Index?

Podemos pensar no **Index** como o "nome" ou "identificador" de cada linha.

- É **imutável** (não pode ser alterado diretamente)
- Garante o **alinhamento dos dados** em operações
- Pode conter valores únicos ou repetidos (embora o ideal seja serem únicos)

---


```python
obj = pd.Series([1,2,3], index=["a", "b", "c"])
obj
```

```python
index_serie = obj.index
print(index_serie)
```

```python
index_serie[1]
```

```python
index_serie[1] = "d" #isso da erro pq o index é imutável
```

```python
#aplicando funcoes
frame = pd.DataFrame(np.random.standard_normal((4, 3)),columns=list("bde"),index=["Utah", "Ohio", "Texas", "Oregon"])
frame
```

```python
def f1(x):
    return x.max() - x.min()

frame.apply(f1) #aplica a função f1 a cada coluna do dataframe, ou seja, calcula a diferença entre o valor máximo e mínimo de cada coluna
```

```python
frame.apply(f1, axis="columns") #aplica a função f1 a cada linha do dataframe, ou seja, calcula a diferença entre o valor máximo e mínimo de cada linha
```

```python
def f2(x):
    return pd.Series([x.min(), x.max()], index=["min", "max"])

a = frame.apply(f2) #aplica a função f2 a cada coluna do dataframe, ou seja, calcula o valor mínimo e máximo de cada coluna e retorna uma série com esses valores
a
```
